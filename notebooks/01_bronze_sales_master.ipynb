```python
csv_path = "/Volumes/retailanalytic/default/retail_landing/sales_data.csv"

bronze_df = spark.read.csv(csv_path, header=True, inferSchema=False)
display(bronze_df.limit(5))
```

```python
import pyspark.sql.functions as F

df = spark.read.table("retailanalytic.default.bronze_sales_master")
silver_df = df.withColumn("TotalAmount", F.col("TotalAmount").cast("decimal(19,4)"))

display(silver_df.select(F.sum("TotalAmount").alias("TotalSales")))
```

```python
display(silver_df.select("SalesDateTime", "IngestedAtUtc").limit(5))
```

```python
silver_typed_df = (
    silver_df
    .withColumn("SalesDateTime", F.col("SalesDateTime").cast("timestamp_ntz"))
    .withColumn(
        "IngestedAtUtc",
        F.concat(F.trim(F.col("IngestedAtUtc")), F.lit("+00:00")).cast("timestamp"),
    )
)

silver_typed_df.printSchema()
display(silver_typed_df.select("SalesDateTime", "IngestedAtUtc").limit(5))
```

```python
silver_df.groupBy("SalesId","LocationId").count().where("count > 1").display()
```

```python
table_name = "retailanalytic.default.silver_sales_master"

silver_df.write.format("delta").saveAsTable(table_name)
```

```python
from pyspark.sql import functions as F

silver = spark.table("retailanalytic.default.silver_sales_master")

gold_df = (
    silver
    .withColumn("LocalDate", F.to_date("SalesDateTime"))
    .withColumn("LocalHour", F.hour("SalesDateTime"))
    .groupBy("LocationId", "LocalDate", "LocalHour")
    .agg(
        F.sum("TotalAmount").alias("TotalSales"),
        F.count("*").alias("ReceiptCount"),
        F.avg("TotalAmount").alias("AvgBasketValue"),
    )
    .orderBy(F.desc("TotalSales"))
)

display(gold_df)
table_name = "retailanalytic.default.gold_sales_master"

gold_df.write.format("delta").saveAsTable(table_name)
display(spark.sql("DESCRIBE DETAIL retailanalytic.default.gold_sales_master"))
display(spark.sql("SELECT * FROM retailanalytic.default.gold_sales_master"))
display(spark.sql("SELECT * FROM retailanalytic.default.silver_sales_master"))
display(spark.sql("SELECT * FROM retailanalytic.default.bronze_sales_master"))
```
